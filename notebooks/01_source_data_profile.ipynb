# SF311 Source Data Profile

This notebook is my first look at the SF311 source data used for OpsPilot.

The goal here is not to clean or model the data yet. I want to understand the schema, check the fields needed for the main KPIs, and identify any obvious data-quality issues before building the warehouse.

Dataset: San Francisco 311 Cases  
Dataset ID: vw6y-z8j6

Checking required columns

```python
required_columns = [
    "service_request_id",
    "requested_datetime",
    "closed_date",
    "updated_datetime",
    "status_description",
    "agency_responsible",
    "service_name",
    "service_subtype",
    "analysis_neighborhood",
    "source"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

print("Missing required columns:", missing_columns)
```

Createing a small profile table

```python
profile = []

for column in required_columns:
    profile.append({
        "column": column,
        "dtype": str(df[column].dtype),
        "missing": df[column].isna().sum(),
        "missing_pct": round(
            df[column].isna().mean() * 100,
            2
        ),
        "unique_values": df[column].nunique(
            dropna=True
        )
    })

profile_df = pd.DataFrame(profile)

profile_df
```

INITIAL FINDINGS

## Initial Findings

I pulled the 10,000 most recent SF311 cases to get a better understanding of the data before building the warehouse.

What I found:

- I did not find any duplicate or missing `service_request_id` values in this sample.
- The data has a good amount of detail, including multiple service categories, subtypes, and neighborhoods.
- A good number of records do not have a `closed_date`, which makes sense because those requests are still open.
- I found 11 records where the `closed_date` is earlier than the `requested_datetime`, so those records will need to be checked before using them in resolution-time calculations.

Things I need to handle later:

- Some fields, especially the date fields, come in as strings and will need to be converted properly in the staging layer.
- Resolution-time calculations will need some data-quality checks so invalid or unusual records do not affect the final metrics.

## Decisions From This Exploration

For the first warehouse version:

- I will use `service_request_id` as the main request key and check uniqueness again when I load the full historical data.
- Date fields will be converted explicitly instead of relying on automatic type detection.
- I will use `analysis_neighborhood` as the main neighborhood field for reporting.
- Historical backlog will be calculated using the request and close dates instead of relying only on the current status.
- I will keep the raw source data unchanged before applying any cleaning or transformations.
- Records with questionable dates will be flagged and reviewed instead of being silently changed.
- Negative resolution times will not be included in valid resolution-time metrics until those records are investigated.

These decisions may change later if the full historical data shows issues that are not visible in this sample.
